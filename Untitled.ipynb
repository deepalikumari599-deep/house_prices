```python
def parse_amount(val):
    if pd.isna(val):
        return np.nan
    val = str(val).strip()
    if "Call for Price" in val:
        return np.nan
    match = re.match(r'([\d.]+)\s*(Cr|Lac)', val, re.IGNORECASE)
    if not match:
        return np.nan
    num, unit = float(match.group(1)), match.group(2).lower()
    return num * 1e7 if unit == "cr" else num * 1e5

df["Amount(in rupees)"] = df["Amount(in rupees)"].apply(parse_amount)
```

```python
def remove_outliers_iqr(df, columns):
    df_clean = df.copy()
    for col in columns:
        Q1 = df_clean[col].quantile(0.25)
        Q3 = df_clean[col].quantile(0.75)
        IQR = Q3 - Q1
        lower = Q1 - 1.5* IQR
        upper = Q3 + 1.5 * IQR
        df_clean = df_clean[(df_clean[col] >=lower) & (df_clean[col] <= upper)]
    return df_clean

new_df = remove_outliers_iqr(df, ['Amount(in rupees)','Carpet Area','Bathroom','Balcony'])
```

```python
param_dist = {
        'n_estimators': [200, 300, 400],
        'max_depth': [6, 8, 10],
        'min_samples_leaf': [1,2,4],
        'min_samples_split': [2,5,10],
    }
search = RandomizedSearchCV(
        RandomForestRegressor(n_jobs=-1, random_state=42),
        param_distributions=param_dist,
        n_iter=15,
        cv=3,
        scoring='r2',
        random_state=42,
        n_jobs=-1,
)
search.fit(X_train, Y_train)
print("Best params:", search.best_params_)
```
